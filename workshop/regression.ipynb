```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_california_housing
```

```python
data = fetch_california_housing()
print(data.DESCR)
```

```python
X = pd.DataFrame(data.data, columns= data.feature_names)
X
```

```python
y = data.target
y
```

```python
plt.figure(figsize= (10,10))
sns.heatmap(X.corr(), annot= True, fmt= ".2f", cmap= 'coolwarm')
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state= 22)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge

linear = LinearRegression()
linear.fit(X_train_scaled, y_train)

lasso_strong = Lasso(alpha= 0.01)
lasso_strong.fit(X_train_scaled, y_train)

ridge_strong = Ridge(alpha= 500)
ridge_strong.fit(X_train_scaled, y_train)

comparison = pd.DataFrame({
    "Feature": X.columns,
    "Linear": linear.coef_,
    "Ridge_strong": ridge_strong.coef_,
    "Lasso_strong": lasso_strong.coef_
})
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
x = np.array(x).reshape(-1,1)
y = np.array(y)

# Create the polynomial features
degree = 3
poly = PolynomialFeatures(degree= degree)
x_poly = poly.fit_transform(x)

# Train model
model = LinearRegression()
model.fit(x_poly, y)

# Create a smooth curve
x_range = np.linspace(min(x), max(x), 100).reshape(-1,1)
x_range_poly = poly.transform(x_range)
y_pred = model.predict(x_range_poly)

# Plot
plt.scatter(x, y, color= 'blue', label= 'Actual Data')
plt.plot(x_range, y_pred, color= 'red', label= f"Polynomial Degree {degree}")
plt.xlabel("X - Values")
plt.ylabel("Y - Values")
plt.legend()
plt.title("Polynomial Regression")
plt.show()
```
