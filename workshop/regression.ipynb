```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
```

```python
data = load_diabetes()
```

```python
print(data.DESCR)
```

```python
X = pd.DataFrame(data.data, columns= data.feature_names)
y = data.target
```

```python
X
```

```python
y
```

```python
X.isna().sum()
```

```python
X.info()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state= 2)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
linear_reg = LinearRegression()
linear_reg
```

```python
linear_reg.fit(X_train, y_train)
```

```python
y_pred = linear_reg.predict(X_test)
```

```python
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
# Evaluate

print(f"R2 Score: {r2_score(y_test, y_pred)}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_pred)}")
print(f"Mean Squared Error: {mean_squared_error(y_test, y_pred)}")
print(f"Root Mean Squared Error: {rmse}")
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
params = {
    'fit_intercept': [True, False],
    'positive': [True, False],
    'copy_X': [True, False]
}
```

```python
grid = GridSearchCV(estimator= LinearRegression(), param_grid= params, cv= 5, verbose= 3)
grid
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_score_
```

```python
grid.best_params_
```

```python
best_model = grid.best_estimator_
best_model
```

```python
y_tuned_pred = best_model.predict(X_test)
y_tuned_pred
```

```python
rmse = np.sqrt(mean_squared_error(y_test, y_tuned_pred))
```

```python
# Evaluation after tuning

print(f"R2 Score: {r2_score(y_test, y_tuned_pred)}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_tuned_pred)}")
print(f"Mean Squared Error: {mean_squared_error(y_test, y_tuned_pred)}")
print(f"Root Mean Squared Error: {rmse}")
```
