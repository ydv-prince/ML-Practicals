```python
import pandas as pd

# Load the dataset
iris = pd.read_csv('iris.csv')

# Display the first 5 rows
print(iris.head())
```

```python
# Summary statistics
print(iris.describe())
```

```python
# Check for missing values
print(iris.isnull().sum())

# Check for duplicates
print(f"Number of duplicates: {iris.duplicated().sum()}")
```

```python
iris = iris.drop_duplicates()
print(f"Shape after removing duplicates: {iris.shape}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(iris, hue='Species', markers=['o', 's', 'D'])
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for i, feature in enumerate(['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']):
    plt.subplot(2, 2, i+1)
    sns.boxplot(x='Species', y=feature, data=iris)
    plt.title(feature)
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Prepare data
X = iris.drop('Species', axis=1)
y = iris['Species']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
```
